```python
import numpy as np
import pennylane as qml

print(f"PennyLane version: {qml.__version__}")

# 1. Initialize the default.qubit simulator with 4 qubits (wires 0, 1, 2, 3)
dev = qml.device("default.qubit", wires=4)


# 2. Define the 4-qubit circuit with Hadamard (H) and CNOT entangling gates
@qml.qnode(dev)
def four_qubit_circuit():
  # Put all 4 qubits into uniform superposition
  for wire in range(4):
    qml.Hadamard(wires=wire)

  # Entangle adjacent qubits in a ring/chain using CNOT gates
  qml.CNOT(wires=[0, 1])
  qml.CNOT(wires=[1, 2])
  qml.CNOT(wires=[2, 3])

  # Return the full 2^4 = 16 complex state vector
  return qml.state()


# 3. Draw the circuit to verify gate placement
print("\nCircuit Diagram:")
print(qml.draw(four_qubit_circuit)())

# 4. Execute and verify the state vector output
state_vector = four_qubit_circuit()
probabilities = np.abs(state_vector) ** 2

print("\nState Vector (16 complex amplitudes):")
for i, amp in enumerate(state_vector):
  basis_state = format(i, "04b")
  print(f"|{basis_state}> : {amp:.4f}  (Prob: {probabilities[i]:.4f})")

# Sanity checks
print(f"\nTotal Probability Sum: {np.sum(probabilities):.6f}")
print(f"Expected amplitude per state (1/sqrt(16)): {1/np.sqrt(16):.4f}")
```
